**A1**: Für eine binomialverteilte Zufallsgröße $Y$ mit den Parametern $n=8$ und $0 < p < 1$
gilt $P(Y=1) = 2 \cdot P(Y=0)$. Berechne $p$.

```python
from sympy import symbols, Eq, solve
import sympy.stats as stats 
p = symbols('p')
n = 8
Y = stats.Binomial('Y', n, p) 
g = Eq(stats.density(Y)(1) , 2*stats.density(Y)(0))
solve(g,p)
```

```python
# c
# Dazu muss C*D = CD sein.
from sympy import symbols, Rational, Eq, solve
x = symbols('x')
D = Rational(2,5)+ x
CD = Rational(2,5)
g = Eq(C * D, CD)
solve(g,x)
```

```python
# c 
# oder: dazu muss DgC = D sein
from sympy import symbols, Rational, Eq, solve
x = symbols('x')
DgC = Rational(3,5)
D = Rational(2,5)+ x
g = Eq(DgC, D)
solve(g,x)
```

```python
''' A5
Nadja und Victor sind Bogenschützen und trainieren oft gemeinsam. Die Wahrscheinlichkeiten in
der abgebildeten Vierfeldertafel beruhen auf ihren bisherigen Trainingsergebnissen. 
V: Victor trifft ins Zentrum, N: Nadja trifft ins Zentrum. Beschrieben wird eine Situation,
in der beide jeweils einen Schuss abgeben Die Ergebnisse N und V sind unabhängig
voneinander. Nadja trifft besser als Victor. Berechne die Wahrscheinlichkeit, dass Nadja ins Zentrum trifft.
'''
from sympy import symbols, Matrix, Eq, solve
x = symbols('x')
tafel = Matrix(3, 3, [0]*9)
 
tafel[0,0] = 0.28
tafel[1,1] = 0.18
tafel[2,2] = 1
tafel[0,2] = x       # einen Wert unbekannt setzen
tafel
```

```python
tafel[0,1] = x - tafel[0,0]
tafel[1,2] = 1 - x
tafel[1,0] = tafel[1,2] - tafel[1,1]
tafel[2,0] = tafel[0,0] + tafel[1,0]
tafel[2,1] = tafel[0,1] + tafel[1,1]
tafel
```

```python
# für die Unabhängigkeit muss gelten: P(V)*P(N) = P(VN)
g = Eq(tafel[2,0] * tafel[0,2], tafel[0,0])
sol = solve(g, x)
sol
```

```python
tafel.subs(x, sol[1])
```

**A7:** 
Für ein Spiel wird ein Behälter mit 100 Kugeln gefüllt. Dafür stehen rote und blaue Kugeln zur Verfügung.
Vor jedem Spiel legt der Spieler die Anzahl der blauen Kugeln im Behälter fest. Anschließend wird dem
Behälter eine Kugel zufällig entnommen. Ist diese Kugel rot, so wird dem Spieler die festgelegte Anzahl
blauer Kugeln in Cent ausgezahlt; ist die Kugel blau, so beträgt die Auszahlung 10 Cent. 
Ermittle, wie der Spieler die Anzahl blauer Kugeln für ein Spiel festlegen muss, damit der 
Erwartungswert der Auszahlung möglichst groß ist.
 

<img src='a7.png' width='401'>

```python
from sympy import symbols, diff, Eq, solve
x = symbols('x')
f = (1 - x/100) * x + x/100 * 10
f1 = diff(f, x)
g = Eq(f1,0)
solve(g,x)
```

a. Berechne die Wahrscheinlichkeit dafür, keine rote Kugel zu ziehen, falls sich in der Urne eine weiße und neun rote Kugeln befinden.

b. Es ist p die Wahrscheinlichkeit dafür, bei einem Zug eine rote Kugel zu ziehen. Die Zufallsgröße X beschreibt die Anzahl der gezogenen roten Kugeln. Bestimme alle Werte für p, für die $P(X=0) = P(X=1)$ gilt.

<img src='a8.png' width='701'>

```python
#a
(S(1)/10)**5
```

```python
#b
import sympy.stats as stats
p = symbols('p')
X = stats.Binomial('X', 5, p)
g = Eq(stats.density(X)(0),stats.density(X)(1))
solve(g,p)
```

**A9**: Ein Unternehmen füllt Honig in Gläser mit der Aufschrift 250g ab. Die Füllmenge $X$ eines Glases
wird als normalverteilt angenommen mit dem Erwartungswert $\mu = 252$ und der Standardabweichung $\sigma = 2$ (alle Angaben in g).

**a.** Ein Glas wird zufällig ausgewählt. Bestimme die Wahrscheinlichkeit dafür dass weniger als 250g in diesem Glas sind.


```python
from scipy.stats import norm
mu = 252
sigma = 2
N = norm(mu,sigma)
x = N.cdf(250)
print(f'{x:.3f}')
```

**b.** Ermittle die Wahrscheinlichkeit dafür, dass die Füllmenge dieses Glases um höchstens 2% von 250 abweicht.

```python
x = N.cdf(250*1.02)-N.cdf(250*0.98)
print(f'{x:.3f}')
```

**c.** Gesucht ist ein Wert $a \ne 245$ mit $P(a \le X \le a+5) = P(245 \le X \le 250)$.

```python
a = 245
x = N.cdf(a+5)-N.cdf(a)    # endet 2 vorm Mittelwert
print(f'{x:.3f}')
a = 254
x = N.cdf(a+5)-N.cdf(a)    # beginnt 2 nach Mittelwert
print(f'{x:.3f}')
```

**d.** Bei einer neuen Abfüllanlage beträgt die Standardabweichung der Füllmenge 1g. Bei ihr kann man durch einen Regler den 
Erwartungswert der Füllmenge mit einer Genauigkeit von 0.1g einstellen. Die Wahrscheinlichkeit dafür, dass ein zufällig ausgewähltes Glas weniger als 250 enthält, soll höchstens 15% betragen. Ermittle, auf welchen Wert der Erwartungswert der Füllmenge mindestens eingestellt werden muss.

```python
from sympy import solve
from sympy.stats import Normal, P
mu = symbols('mu') 
X = Normal('X', mu, 1)
g = Eq(P(X < 250),0.15)
solve(g,mu)
```

Der nächste aufgerundete Wert in 0.1 Schritten ist also $251.1$

Mit scipy hätte man das in einer Schleife lösen können:

```python
from scipy.stats import norm
mu = 200                  # Erwartungswert ganz niedrig einstellen
sigma = 1
N = norm(mu,sigma)
w = N.cdf(250)
while w > 0.15:
    mu+=0.1               # Erwartungswert um 0.1 erhöhen
    N = norm(mu,sigma)
    w = N.cdf(250)
print(f'{mu:.1f}') 
```

**h**: Die Gläser werden in Kartons abgepackt und an Lebensmittelgeschäfte ausgeliefert. Jeder Karton enthält 30 Gläser. Ein Kunde nimmt drei Gläser aus dem Karton und hofft, mit einer Wahrscheinlichkeit von mindestens 50% mindestens einen Gutschein zu erhalten. Ermittle die Anzahl der Gläser mit Gutschein, die sich dafür mindestens in dem Karton befinden müssen.

```python
# h - Ziehen ohne Zurücklegen folgt der hypergeometrischen Verteilung
from scipy.stats import hypergeom
N = 30     # Gesamtzahl der Objekte
K = 0      # Anzahl günstiger Objekte
n = 3      # Anzahl der gezogenen Objekte
H = hypergeom(N, K, n)

k = 0      # Anzahl gezogener günstiger Objekte
prob = 1 - H.pmf(k)

while prob < 0.5:
    K+=1
    H = hypergeom(N, K, n)
    prob = 1 - H.pmf(k)
    
print(f'Bei {K} Gläsern mit Gutschein: P={prob:.4f}')
```

**A10**:
Bei einem Glücksrad gibt es drei farbige Sektoren. Beim einmaligen Drehen beträgt die Wahrscheinlichkeit für rot 60%, für blau 30% und für grün 10%. Das Glücksrad wird 20-mal gedreht.

**a.** Bestimme die Wahrscheinlichkeit des Ereignisses: 'Genau 15-mal erscheint rot'.

```python
from scipy.stats import binom
r, b, g = 0.6, 0.3, 0.1
X = binom(20,r)
print(round(X.pmf(15),3))
```

**b**: Bestimme die Wahrscheinlichkeit des Ereignisses: 'Bei den ersten 10 Durchgängen erscheint genau zweimal blau, insgesamt erscheint höchstens fünfmal blau'

```python
X = binom(10,b)
p1 = X.pmf(2)
X = binom(10,b)
p2 = X.cdf(3)
print(round(p1*p2,3))
```

**c**: Bei einem Glücksspiel darf man für einen Einsatz von 6€ das Glücksrad zweimal drehen. Wenn dabei genau
einmal rot erscheint, dann erhält man einen bestimmten Auszahlungsbetrag. Wenn zweimal rot erscheint, dann erhält man
das Siebenfache dieses Auszahlungsbetrags. Andernfalls erfolgt keine Auszahlung. Das Spiel ist fair. Bestimme den Auszahlungsbetrag
für den Fall, dass genau einmal rot erscheint.


```python
from sympy import symbols, Eq, solve
x = symbols('x')    # Auszahlungsbetrag
einsatz = 6
X = binom(2,r)

E = x * X.pmf(1) + 7 * x * X.pmf(2) - einsatz   # Erwartunswert Gewinn
g = Eq(E,0)                                     # Spiel ist fair
sol = solve(g,x)
sol
```

**A11**: Bei einem Spielautomaten wird vermutet, dass die Gewinnwahrscheinlichkeit $p$ großer als $10%$ ist. Die Vermutung wird mithilfe eines Hypothesestests mit dem Stichprobenumfang von $n=200$ und einem Signifikanzniveau von $5\%$ getestet. Als Nullhypothese wird 
$H_0: p \ge 0.1$ gewählt. 

**a.:** Formuliere die Entscheidungsregel

Mit dem folgenden Code ist sympy überfordert, deswegen nutzen wir eine Schleife.

```python
'''
k = symbols('k')
X = stats.Binomial('X', 200, 0.1) 
ug = stats.P(X <= k) <= 0.05
solve_univariate_inequality(ug,k)
''';
```

```python
from scipy.stats import binom
n = 200
p = 0.1
alpha = 0.05

# größtes k mit P(X <= k) <= alpha
k = 0      
while binom.cdf(k,n,p) <= alpha:
    k += 1   
k_crit = k-1         # das letzte -1 muss wieder zurückgenommen werden

print(f"Ablehnungsbereich: 0..{k_crit}")
```
