### Abi-Stochastik Cheetsheet

```python
import math
import numpy as np
from scipy.stats import binom, norm

X = binom(8,0.5)      # X binomialverteilt mit n=8, p=0.5
X.pmf(3)              # P(X=3)
X.cdf(3)              # P(X<=3)
X.mean(), X.std(), X.var()   #  Erwartungswert, Standardabweichung, Varianz

X = norm(0,1)         # Normalverteilung mit mu = 0, sigma = 1
X.pdf(3)              # y-Wert der Glockenkurve
X.cdf(3)              # P(X<=3)
X.ppf(0.3)            # für welches k gilt P(X<=k) = 0.3  (cdf-invers)

math.comb(8,2)        # Binomialkoeffizient 8 über 2

tafel = np.zeros(9).reshape(3,3)   # Vierfeldertafel
tafel[2,2] = 1
```
