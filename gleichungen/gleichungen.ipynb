```python
from sympy import symbols, Eq, solv, binomial
```

```python
# eine einfache Gleichung
x = symbols('x')
g = Eq(2*x + 3, 7)
sol = solve(g, x)
print(sol)
```

```python
# Zwei Gleichungen mit zwei Unbekannten
x, y = symbols('x y')

g1 = Eq(2*x + y, 5)
g2 = Eq(x - y, 1)

# Gleichungssystem lösen
sol = solve((g1, g2), (x, y))

print(sol)
```

```python
from sympy import symbols, Eq, solve, binomial, simplify

# Variable definieren
x = symbols('x')

# Gleichung mit Binomialkoeffizient
gleichung = Eq(binomial(x, 2), 10)

# Gleichung lösen
lösung = solve(gleichung, x)

print(lösung)
```

```python
p = symbols('p')
L = binomial(8,1)*p*(1-p)**7
R = binomial(8,0)*(1-p)**8
g = Eq(L,2*R)
solve(g,p)
    
```

```python
import numpy as np

# -1 steht für "unbekannt"
tafel = np.array([
    [20, 30],   # Zeile A
    [10, -1]    # Zeile ¬A
])

# Gesamtzahl
n = 100

# fehlenden Wert berechnen
tafel[1,1] = n - np.sum(tafel[0]) - tafel[1,0]

# Summen berechnen
zeilensummen = np.sum(tafel, axis=1)
spaltensummen = np.sum(tafel, axis=0)
gesamt = np.sum(tafel)

print("Vierfeldertafel:")
print(tafel)
print("Zeilensummen:", zeilensummen)
print("Spaltensummen:", spaltensummen)
print("Gesamt:", gesamt)
```
