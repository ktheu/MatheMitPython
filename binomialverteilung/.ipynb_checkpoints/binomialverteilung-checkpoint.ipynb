### Wahrscheinlichkeitsrechnung mit der Binomialverteilung

[Video](https://youtu.be/hGemn8AoZM8)

#### Grundbegriffe

Eine Zufallsvariable $X$ ist eine Funktion die jedem Ergebnis eines Zufallsexperiments eine Zahl zuordnet.

Ein Bernoulli-Experiment ist ein Zufallsexperiment mit 2 möglichen Ergebnissen.

Die Zufallsvariable $X$ ist binomialverteilt mit den Parametern $n$ und $p$, $X \sim  B(n,p)$, wenn sie die 
die Anzahl der Erfolge bei $n$ unabhängigen Bernoulli-Experimenten beschreibt, bei denen jeder Versuch mit der gleichen Wahrscheinlichkeit 
$p$ erfolgreich ist.

Beispiel:

    Zufallsexperiment: Würfle 8-mal. 
    X = Anzahl der Male, bei denen eine gerade Zahl (2, 4, 6) kommt.

    Zufallsexperiment: Wirf eine faire Münze 8-mal
    Y = Anzahl der Male, bei denen Kopf kommt

<img src='bild01.png' width='400'>

$X$ und $Y$ sind beide $B(8,0.5)$ verteilt. $X$ und $Y$ sind trotzdem verschiedene Zufallsvariablen, denn als Funktionen haben sie unterschiedliche Definitionsbereiche.

Für Aufgaben zur Binomialverteilung benutzen wir *scipy.stats*. <br>
*scipy* führt numerische Berechungen durch, ist schnell und bietet viele Funktionen.


```python
import matplotlib.pyplot as plt
import numpy as np
import math
from scipy.stats import binom
```

##### Binomialkoeffizient

<img src='bild02.png' width='500'>

Bedeutungen:
- Anzahl verschiedener Teilmengen der Größe 5 aus einer Menge mit 8 Elementen: $\binom{8}{5}$
- Anzahl verschiedener Teams der Größe 11 aus einer Menge von 20 Personen:  $\binom{20}{11}$
- Anzahl verschiedener 5 Karten auf der Hand aus einer Menge von 52 Karten  $\binom{52}{5}$

```python
math.comb(8,3) 
```

##### Wahrscheinlichkeit berechnen: $P(X=k)$

*pmf* steht für *probability mass function*

```python
# P(X = 3)
X = binom(8,0.5)     #  X binomial verteilt mit n=8, p=0.5
X.pmf(3)
print(X.pmf(3))
```

##### Wahrscheinlichkeit berechnen: $P(X \le k)$

*cdf* steht für *cumulative density function*

```python
# P(X <= 3)
X = binom(8,0.5)
X.cdf(3)
```

##### Erwartungswert, Varianz, Standardabweichung

```python
X = binom(8, 0.5)
print(f'Erwartungswert = {X.mean()}')
print(f'Standardabweichung = {X.std()}')
print(f'Varianz = {X.var()}')
```

##### Die Wahrscheinlichkeiten grafisch darstellen

```python
n, p = 8, 0.5
X = binom(n,p)
xx = np.arange(n + 1)
yy = X.pmf(xx)

plt.subplots(figsize=(5,3))    
plt.bar(xx, yy)
plt.xlabel('Anzahl Erfolge')
plt.ylabel('Wahrscheinlichkeit')
plt.title(f'Binomialverteilung B({n}, {p})')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

##### Die kumulative Verteilungsfunktion

```python
n, p = 8, 0.5
X = binom(n,p)
xx = np.arange(n + 1)
yy = X.cdf(xx)

plt.subplots(figsize=(5,3))    
plt.step(xx, yy, label='CDF (Treppenfunktion)', color='blue')
 
# Zusatz: Gitter und Achsenbeschriftung
plt.xlabel('Anzahl Erfolge k')
plt.ylabel('P(X ≤ k)')
plt.title(f'Kumulative Verteilungsfunktion der Binomialverteilung B({n}, {p})')
plt.legend()
plt.yticks(np.arange(0, 1.1, step=0.1))
plt.grid(True)
plt.ylim(0, 1.05)  # y-Achse bis etwas über 1, damit obere Stufe gut sichtbar ist
plt.show()


```

##### Binomialverteilte Daten erzeugen


```python
n, p = 8, 0.5
anzahl = 100
X = binom(n,p)
daten = X.rvs(anzahl)
daten
```

```python
# Plot histogram
plt.subplots(figsize=(5,3))    
plt.hist(daten, bins=range(n+2), align='left', rwidth=0.8)
plt.xlabel('Anzahl Erfolge')
plt.ylabel('Häufigkeit')
plt.title(f'{anzahl} binomial verteilte Daten (n={8}, p={0.5})')
plt.xticks(range(n+1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
 
plt.show()
```
