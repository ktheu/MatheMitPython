```python
from sympy import symbols, Eq, solve, binomial, Rational, Matrix, diff
```

```python
'''
P 2021
Für eine binomialverteilte Zufallsgröße Y mit den Parametern n=8 und 0 < p < 1
gilt P(Y=1) = 2 * P(Y=0). Berechne p.

Lösung: 1/5
'''
p = symbols('p')
L = binomial(8,1)*p*(1-p)**7
R = binomial(8,0)*(1-p)**8
g = Eq(L,2*R)
solve(g,p)
```

```python
'''
W 2021
Eine Urne enthält 100 Kugeln, davon sind 40 weiß, davon n weiße Kugeln 1 < n < 99.
Berechne n, wenn die Wahrscheinlichkeit, dass zwei weiße Kugeln gezogen werden,
beim Ziehen mit Zurücklegen um 2% größer ist als beim Ziehen ohne Zurücklegen.
''';
```

```python
n = symbols('n')
p_mit = (n / 100) * (n / 100)
p_ohne = (n / 100) * ((n - 1) / 99)
g = Eq(p_mit, 1.02 * p_ohne)  
solve(g,n) 
```

```python
'''
Nadja und Victor sind Bogenschützen und trainieren oft gemeinsam. Die Wahrscheinlichkeiten in
der abgebildeten Vierfeldertafel beruhen auf ihren bisherigen Trainingsergebnissen. 
V: Victor trifft ins Zentrum, N: Nadja trifft ins Zentrum. Beschrieben wird eine Situation,
in der beide jeweils einen Schuss abgeben Die Ergebnisse N und V sind unabhängig
voneinander. Nadja trifft besser als Victor. Berechne die Wahrscheinlichkeit, dass Nadja ins Zentrum trifft.
''';
```

```python
x = symbols('x')
tafel = Matrix(3, 3, [0]*9)

# bekannte Werte
tafel[0,0] = 0.28
tafel[1,1] = 0.18
tafel[2,2] = 1
tafel
```

```python
tafel[0,2] = x               # einen Wert unbekannt setzen
tafel[0,1] = x - tafel[0,0]
tafel[1,2] = 1 - x
tafel[1,0] = tafel[1,2] - tafel[1,1]
tafel[2,0] = tafel[0,0] + tafel[1,0]
tafel[2,1] = tafel[0,1] + tafel[1,1]
tafel
```

```python
g = Eq(tafel[2,0] * tafel[0,2], tafel[0,0])
sol = solve(g, x)
```

```python
tafel_mit_loesung = tafel.subs(x, sol[1])
tafel_mit_loesung
```

```python
'''
Für ein Spiel wird ein Behälter mit 100 Kugeln gefüllt. Dafür stehen rote und blaue Kugeln zur Verfügung.
Vor jedem Spiel legt der Spieler die Anzahl der blauen Kugeln im Behälter fest. Anschließend wird dem
Behälter eine Kugel zufällig entnommen. Ist diese Kugel rot, so wird dem Spieler die festgelegte Anzahl
blauer Kugeln in Cent ausgezahlt; ist die Kugel blau, so beträgt die Auszahlung 10 Cent. 
Ermitteln Sie, wie der Spieler die Anzahl blauer Kugeln für ein Spiel festlegen muss, damit der 
Erwartungswert der Auszahlung möglichst groß ist.
'''
x = symbols('x')
f = (1 - x/100) * x + x/100 * 10
f1 = diff(f, x)
g = Eq(f1,0)
solve(g,x)
```

```python
import sympy.stats as stats
p = symbols('p')
n = 5

X = stats.Binomial('X', n, p)

# Compute the PMF
L = stats.density(X)(0)
R = stats.density(X)(1)
g = Eq(L,R)
solve(g,p)
```
