```python
# f
mu = 600.5
mu1= 600.8
sigma = 0.23
N = norm(mu,sigma)
N1 = norm(mu1,sigma)

fig, ax = plt.subplots(figsize=(10,4))
X = np.linspace(599.5,601.5,400)
Y = N.pdf(X)
Y1 = N1.pdf(X)
 
ax.plot(X,Y)
ax.plot(X,Y1)
plt.show()
```

```python
# f
mu = 600.5
sigma = 0.23
sigma1 = 0.2
N = norm(mu,sigma)
N1 = norm(mu,sigma1)

fig, ax = plt.subplots(figsize=(10,4))
X = np.linspace(599.5,601.5,400)
Y = N.pdf(X)
Y1 = N1.pdf(X)
 
ax.plot(X,Y)
ax.plot(X,Y1)
plt.show()
```

```python
# g
def p(n):
    tmp = 1
    for i in range(1,n):
        tmp*=(n-i)/n
    return tmp
        
n = 1
while p(n) >= 0.01:
    n+=1
print(n)

```

```python
#e
'''
Wenn die Entscheidungsregel links liegt, wird also ein linksseitiger Test gemacht,
d.h. die Nullhypothese behauptet, die Wahrheit liegt rechts.
H0: p >= 2/3
'''
X = binom(500,2/3)
X.cdf(318)     # Ist nicht im Ablehnungsbereich
```

```python
X.cdf(317)    # Ist im Ablehnungsbereich
```
