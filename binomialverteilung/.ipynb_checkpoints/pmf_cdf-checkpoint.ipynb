### pmf und cdf der Binomialverteilung

#### scipy

$X \sim B(8,0.5)$

Wenn wir zu gegebenen n und p Werte berechnen wollen, nutzen wir *binom* aus *scipy.stats*

```python
from scipy.stats import binom
n = 8
p = 0.5
X = binom(n,p)
```

$P(X = 3)$, $P(X \le 3)$

```python
k = 3
print(X.pmf(k))
print(X.cdf(k))
```

#### sympy - pdf



Wenn in unserer Aufgabenstellung n oder p oder k Unbekannte sind, nutzen wir sympy.

```python
import sympy.stats as stats
from sympy import symbols, S
n = 8
p = 0.5
X = stats.Binomial('X', n, p) 
```

```python
k = 3
stats.density(X)(k)
```

```python
# Falls p unbekannt
p = symbols('p')
n = 8
k = 3
X = stats.Binomial('X', n, p) 
stats.density(X)(k)
```

```python
# Einsetzen
s = stats.density(X)(k)
s.subs({'p':0.5})
```

```python
# Falls n, p, k unbekannt
n, p, k = symbols('n p k')
X = stats.Binomial('X', n, p) 
stats.density(X)(k)
```

```python
# Einsetzen
s = stats.density(X)(k)
s.subs({'p':0.5, n:8, k:3})
```

#### sympy - cdf

```python
n = 8
p = 0.5
X = stats.Binomial('X', n, p)
k = 3
stats.P(X <= k)
```

```python
# Falls p unbekannt
p = symbols('p')
n = 8

X = stats.Binomial('X', n, p)
k = 3
stats.P(X <= k)
```
