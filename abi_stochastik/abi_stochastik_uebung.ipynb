**a.** Bestimme die Wahrscheinlichkeit des Ereignisses: 'Genau 66 der verkauften Patronen sind mit schwarzer Tinte gefüllt':

```python
X = binom(100,0.65)
ausgabe(X.pmf(66))
```

**h.** Betrachtet wrid die für $x \ge 0$ definierte Funktion f mit $f(x) = 0.25 \cdot e^{-0.25x}$. Weise nach, dass $f$ eine
Dichtefunktion über ihrem Definitionsbereich ist.

```python
x = symbols('x')
f = 0.25 * exp(-0.25 * x)
F = integrate(f, x)
print("Stammfunktion:", F)
```

```python
integrate(f, (x, 0, oo))
```

**i**: Die Zeitdauer in Stunden zwischem dem Eingang einer Bestellung im Onlineshop und dem
Versand der Ware kann modellhaft durch die stetige Zufallsgröße mit Dichtefunktion $f$ beschrieben werden.
Mit einer Wahrscheinlichkeit von 85% wird eine Ware innerhalb von t Stunden nach Eingang der Bestellung versandt. Bestimme den Wert von t.

```python
t = symbols('t') 
L = integrate(f, (x, 0, t))   
g = Eq(L,0.85)
sol = solve(g,t)
ausgabe(sol[0])
```

```python
n = 1               # Anzahl Mitspieler
p = 1/3+1/4
X = binom(n,p)      # Alle Mitspieler haben Werte 3 oder 4 
while X.pmf(n) > 0.02:
    n+=1
    X = binom(n,p)  
    X.pmf(n)
print(n)
```
