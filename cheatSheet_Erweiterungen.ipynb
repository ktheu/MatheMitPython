```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
# Dichtefunktion der Exponentialverteilung mit lambda = 2
from scipy.stats import expon
lambda_ = 2
E = expon(scale=1/lambda_)    
fig, ax = plt.subplots(figsize=(4,3))
X = np.linspace(0,8,400)
Y = E.pdf(X)
ax.plot(X,Y)
plt.show()
```

```python
# Histogramm einer Hypergeometrischen Verteilung (Ziehen ohne Zurücklegen)
from scipy.stats import hypergeom
N = 30     # Gesamtzahl der Objekte
K = 6      # davon K Treffer
n = 3      # Anzahl der gezogenen Objekte

H = hypergeom(N,K,n)
X = np.arange(n+1)
Y = H.pmf(X)
fig, ax = plt.subplots(figsize=(4,3))
X = np.arange(n+1)
Y = H.pmf(X)
ax.bar(X,Y)
plt.xlabel("Anzahl gezogener Treffer")
plt.show()
```
