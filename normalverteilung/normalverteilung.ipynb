### Normalverteilung

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm
```

```python
N = norm(0,1)               # Normalverteilung(0,1)
print(f'mean={N.mean()}')
print(f'stdabweichung={N.std()}')
```

```python
fig, ax = plt.subplots()
X = np.linspace(-3,3,400)
Y = N.pdf(X)
ax.plot(X,Y)
plt.show()
```

#### Normalverteilte Daten

```python
r = N.rvs(size=10000)
```

```python
fig, ax = plt.subplots()
X = np.linspace(-3,3,400)
Y = N.pdf(X)
ax.plot(X,Y)
ax.hist(r, bins='auto', density=True, alpha=0.3)
plt.show() 
```

#### pdf und cdf


```python
N = norm(0,1)   
print(N.pdf(-1))            # der Funktionswert
print(N.cdf(-1))            # die Fläche bis zu diesem Funktionswert
```

```python
print(N.pdf(0)) 
print(N.cdf(0))
```

```python
print(N.cdf(1)-N.cdf(-1))
print(N.cdf(2)-N.cdf(-2))
```

#### Beispiel

Die Körpergröße eines 17-jährigen.

```python
from matplotlib.ticker import MultipleLocator 
N = norm(178, 7.5)
fig, ax = plt.subplots(figsize=(10,4))
X = np.linspace(155,200,400)
Y = N.pdf(X)
ax.xaxis.set_major_locator(MultipleLocator(5))
ax.plot(X,Y)
ax.axvline(N.mean(),color='black',linestyle=':')

ax.fill_between(X, Y, where=(170 <= X) & (X <= 180),
                color='green', alpha=0.3)
plt.ylim(0,0.055)
ax.set_title('Körpergröße eines 17-jährigen')
plt.grid()
plt.show() 
```

Die Wahrscheinlichkeit, dass ein 17-jähriger zwischen 170 und 180 groß ist:

```python
print(f'{N.cdf(180)-N.cdf(170):.3f}')
```
